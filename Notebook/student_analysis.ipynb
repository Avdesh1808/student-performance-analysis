```python
import numpy as np

student_data["result"] = np.where(
    student_data["final_score"] >= 40,
    "Pass",
    "Fail"
)
Fail_student = (list(np.where(student_data["result"]=="Fail"))) # check Failed students ID
for item in Fail_student:
    print(student_data.loc[item,"student_id"])
```

```python
Pass_student = (list(np.where(student_data["result"]=="Pass")))
for item in Pass_student:
    print(student_data.loc[item,"student_id"])
```
